```python
#  Data Cleaning

#  Remove extra spaces from column names
df.columns = df.columns.str.strip()
print("true")
```

```python
# Checking missing values

print(df.isnull().sum())
```

```python
#  Calculate the missing value percentage
missing_percentage = (df.isnull().sum()/len(df))*100
print(missing_percentage)
```

```python
# Remove rows with missing values
df=df.dropna().copy()

# Remove duplicate records
df =df.drop_duplicates().copy()

print("missing values after cleaning:")
print(df.isnull().sum())

print("Dataset shape:",df.shape)
```

```python
# Convert date into datetime format:

df['Date'] = pd.to_datetime(
    df["Date"],
    dayfirst=True,
    errors='coerce'
)

df = df.dropna(subset=['Date']).copy()

# Sort by date
df=df.sort_values("Date")
df
```

```python
df.head()
```

```python
#  Clean category labels:

df['Region']=df["Region"].str.strip()
df["Area"]=df["Area"].str.strip()
df["Freauency"]=df["Frequency"].str.strip()

df["Area"].unique()
```

```python
#  Feature Engineering - Create additional columns to make my analysis more useful
# Extract month and year
```

```python
# Extract month and year
df['Year'] = df['Date'].dt.year
df['Month'] = df['Date'].dt.month_name()
df['Month_Number'] = df['Date'].dt.month

# Create year-month column
df['Year_Month'] = df['Date'].dt.to_period('M').astype(str)

df.head()
```

```python
#  Statastical analysis
# 1)What is the average unemployment rate?

average_unemployment = df[
    'Estimated Unemployment Rate (%)'
].mean()

print("Average Unemployment Rate:",
      round(average_unemployment, 2), "%")
```

```python
#  Calculate additonal analysis
print("maximum Unemployment:",
      df['Estimated Unemployment Rate (%)'].max())

print("Minimum Unemployment:",
     df["Estimated Unemployment Rate (%)"].min())

print("Median Unemployment:",
     df["Estimated Unemployment Rate (%)"].median())

print("Standard Deviation Unemployment:",
     df["Estimated Unemployment Rate (%)"].std())
```

```python
#  State-Wise Unemployment analysis
# Which states have the highest unemployment rate?

state_unemployment = (
    df.groupby('Region')['Estimated Unemployment Rate (%)']
    .mean()
    .sort_values(ascending=False)
)

print(state_unemployment)
```

```python
# Visualize the top 10 states:

plt.figure(figsize=(12, 6))

state_unemployment.head(10).plot(
    kind='bar',
    color='tomato'
)

plt.title('Top 10 States with Highest Average Unemployment')
plt.xlabel('State')
plt.ylabel('Unemployment Rate (%)')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
# Which states have the lowest unemployment rate?

plt.figure(figsize=(12,10))

state_unemployment.tail(10).plot(
    kind='bar',
    color="Red",
)

plt.title("Top 10 states with Highest Average Unemployment:")
plt.xlabel("State")
plt.ylabel("Unemployment rate (%)")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
bottom_10 = state_unemployment.sort_values().head(10)

plt.figure(figsize=(12, 6))

sns.barplot(
    x=bottom_10.values,
    y=bottom_10.index
)

plt.title('Top 10 States with Lowest Average Unemployment')
plt.xlabel('Unemployment Rate (%)')
plt.ylabel('State')
plt.show()
```

```python
# Rural Vs Urban Analysis
# Is urban Unemployment higher than rural unemployment

area_analysis = df.groupby('Area')[
    "Estimated Unemployment Rate (%)"
].mean()

print(area_analysis)
```

```python
# Comparison chart

plt.figure(figsize=(8, 5))

sns.barplot(
    x=area_analysis.index,
    y=area_analysis.values
)

plt.title('Rural vs Urban Unemployment Rate')
plt.xlabel('Area')
plt.ylabel('Average Unemployment Rate (%)')

plt.show()
```

```python
#  Which state have the highest urban and rural unemploument
area_state = df.groupby(
    ["Region",'Area']
)["Estimated Unemployment Rate (%)"].mean().reset_index()

plt.figure(figsize=(14,7))

sns.barplot(
    data=area_state,
    x="Region",
    y="Estimated Unemployment Rate (%)",
    hue = "Area"
)

plt.title("State-wise rural vs urban unemployment")
plt.xticks(rotation=90)
plt.tight_layout()
plt.show()
```

```python
# TIme-Series analysis
# How does unemployment change over time

monthly_trend  = df.groupby('Year_Month')[
    "Estimated Unemployment Rate (%)"
].mean()

plt.figure(figsize=(14,6))

plt.plot(
    monthly_trend.index,
    monthly_trend.values,
    marker='o'
)

plt.title("Monthly Unemployment Trend in India")
plt.xlabel('Month')
plt.ylabel("Average Unemployment Rate (%)")
plt.tight_layout()
plt.show()
```

```python
# WHich month recorded the Highest average unemployent

month_analysis = df.groupby('Month')[
    "Estimated Unemployment Rate (%)"
].mean().sort_values(ascending=False)

print(month_analysis)
```

```python
# Year-wise comparison
# How did unemployment change between 2019 and 2020
year_analysis = df.groupby("Year")[
     'Estimated Unemployment Rate (%)'
].mean()

print(year_analysis)

year_analysis.plot(
    kind='bar',
    figsize=(8,6),
    title='Year-wise Average Unemployment Rate'
)

plt.xlabel('Year')
plt.ylabel('Unemployment Rate (%)')
plt.show()
```

```python
#  Correlation Analysis
# Is there a relationship between uneployment and labour participation

correlation = df[
    [
        "Estimated Unemployment Rate (%)",
        "Estimated Labour Participation Rate (%)",
        "Estimated Employed"
    ]
].corr()

print(correlation)

plt.figure(figsize=(8,6))

sns.heatmap(
    correlation,
    annot=True,
    cmap='coolwarm',
    fmt='.2f'
)

plt.title("Correlation Between Employment Indicators")
plt.show()
```

```python
#  Employment analysis
# WHich state have the highest estimated employment

employment_analysis = df.groupby('Region')[
    "Estimated Employed"
].mean().sort_values(ascending=False)

print(employment_analysis.head(10))
```

```python
employment_analysis.head(10).plot(
    kind='bar',
    figsize=(12,6)
)

plt.title('Top 10 States y average Estimated Employment')
plt.xlabel('State')
plt.ylabel('Estimated Empoyed')
plt.show()
```

```python
#  Distribution and outlier analysis
# How are unemployment rates distributed

plt.figure(figsize=(10,5))

sns.histplot(
    df['Estimated Unemployment Rate (%)'],
    bins=30,
    kde=True
)

plt.title("Distibution of Unemployment Rate")
plt.xlabel("Unemployment Rate (%")
plt.show()
```

```python
# Are there any outliers

plt.figure(figsize=(10,5))

sns.boxplot(
    x=df["Estimated Unemployment Rate (%)"]
)

plt.title('Unemployment Rate Outlier Analysis')
plt.show()
```

```python
# State with the Highest changes
# which states experienced the highest increase or decrease

state_monthly = df.groupby(
    ['Region', 'Date']
)['Estimated Unemployment Rate (%)'].mean().reset_index()

state_change = state_monthly.sort_values(
    ['Region', 'Date']
).groupby('Region').agg(
    First_Rate=('Estimated Unemployment Rate (%)', 'first'),
    Last_Rate=('Estimated Unemployment Rate (%)', 'last')
)

state_change['Change'] = (
    state_change['Last_Rate'] - state_change['First_Rate']
)

print("Largest increases:")
print(state_change.sort_values('Change', ascending=False).head(10))

print("Largest decreases:")
print(state_change.sort_values('Change').head(10))
```

```python
# COVID-19 period analysis

df['Period'] = np.where(
    df['Date'] < '2020-03-01',
    'Before COVID-19',
    'During COVID-19'
)

covid_analysis = df.groupby('Period')[
    'Estimated Unemployment Rate (%)'
].mean()

print(covid_analysis)

covid_analysis.plot(
    kind='bar',
    figsize=(8, 5),
    title='Unemployment Before and During COVID-19'
)

plt.ylabel('Average Unemployment Rate (%)')
plt.show()
```
